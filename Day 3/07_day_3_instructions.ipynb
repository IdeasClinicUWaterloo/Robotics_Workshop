# Day 3 - Competition Instructions

Now that you have practiced collecting data and tuning your model, it is time to compete!

Your goal is to train a model that can accurately distinguish between the motor's idle and stall conditions using the data that we have provided.

We recommend creating a new project for this challenge.

## Important Notes About the Data

For the competition, we have provided both training data (`training.zip`) and test data (`test.zip`) for you to upload to Edge Impulse. All data is labelled as either idle or stall. Idle data involves the motor running unobstructed for 10 seconds. Stall data includes ***at least one*** stall somewhere during the 10 second run of the motor.

Note that before training your model, you are responsible for cleaning **both the training and test data**.

You can use as much or as little of the provided training data as you like. **However**, you must only use the test data provided by us for testing your final model. Do not collect additional data or reassign training data to test data. You must also use **all** of the test data we provide. This will allow for a fairer competition.

About 30 minutes before the end of the workshop, we will check your model. At the end of the workshop, we will announce the group whose model has the **highest accuracy in the Model Testing tab**.

## How to Import Data into Edge Impulse

The provided data is in the format of JSON files exported from Edge Impulse. To upload them into Edge Impulse:

1) Download the provided `training.zip` and `test.zip` files and extract them.
2) Navigate to the **Data Acquisition** tab within your Edge Impulse project and click on the **upload data** button.

<img src = "Assets/Upload_Data.png" width="60%" >

3) Ensure that the **upload mode** is set to **select a folder**.
4) Click on **choose files** and select either the extracted training folder or test folder from your computer.
5) Based on whether you are uploading the training or test data, select the appropriate category.
6) We recommend allowing Edge Impulse to **infer the label from the filename**. You can then change any labels that need to be altered while cleaning the data.
