# Edge Impulse Setup

[Edge Impulse](https://studio.edgeimpulse.com/login) is the platform we will use in this workshop to build, train, and refine machine learning models for motor fault prediction. Please click on the hyperlink above and create an Edge Impulse account.

Once you have signed up, log into your account and create a new project. Keep the default settings and enter a name for the project.

![Create_Project.png](Assets/Create_Project.png)

Next, we need to set up the Edge Impulse Data Forwarder. This is a command-line tool that allows us to send sensor data from a device to Edge Impulse Studio, where the data can be collected, cleaned, and labelled to create training datasets for our machine learning models. Follow the steps below to install the required packages.

## Install Python 3
1) On Windows, open Command Prompt and run `python --version`. On Mac, open Terminal and run `python3 --version`. If you have Python 3 installed, you will see something like `Python 3.12.6`. Note that on Mac, you may have to install command-line developer tools in order to run this command.
2) If you do not have [Python 3](https://www.python.org/downloads/release/python-3150rc2/) installed, click on the hyperlink, scroll down, and download the installer.
3) Open the installer and follow the steps on screen.
4) Once the installation has completed, close and reopen Command Prompt/Terminal. Then, run `python --version` or `python3 --version` again to check that it has been properly installed.

## Install Node.js
1) Open Command Prompt (on Windows) or Terminal (on Mac) and run `node --version`. If you have Node.js installed, you will see something like `v24.21.0`
2) If you do not have [Node.js](https://nodejs.org/en/download/current) installed, click on the hyperlink, scroll down to the prebuilt Node.js section, select your operating system and architecture, and download the installer. 

    *NOTE: to check your system architecture on Windows, open Settings > System > About. Then, look at System Type. On Mac, open the Apple menu, select About This Mac, and look at the Chip item. Apple silicon chips use ARM64 architecture and Intel processors use x86_64.*

3) Open the installer and follow the steps on screen. Ensure that you check the box next to "Automatically install the necessary tools...".
4) Once the installation has completed, close and reopen Command Prompt/Terminal. Then, run `node --version` again to check that Node.js has been properly installed.

## Change the npm Default Directory
1) On Windows, open PowerShell and run `Set-ExecutionPolicy RemoteSigned -Scope CurrentUser`.
2) On Windows, in PowerShell run the following commands one after another to change npm's default directory:

    `mkdir ~/.npm-global`

    `npm config set prefix '~/.npm-global'`

    `echo 'export PATH=~/.npm-global/bin:$PATH' >> ~/.profile`

    OR

    On Mac, in Terminal run the following commands one after another to change npm's default directory:

    `mkdir ~/.npm-global`

    `npm config set prefix '~/.npm-global'`

    `echo 'export PATH=~/.npm-global/bin:$PATH' >> ~/.zprofile`

## Download Visual Studio Build Tools - WINDOWS ONLY
1) If you do not have [Visual Studio Build Tools](https://visualstudio.microsoft.com/downloads/?q=build+tools) installed, click on the hyperlink, scroll down to **02 Tools for Visual Studio**, expand it, and download **Build Tools for Visual Studio 2026**. 

2) Open the installer and follow the instructions on screen.
3) Once **Visual Studio Installer** has opened, ensure the box beside **Desktop development with C++** is checked. Then, select **Install while downloading** and click **Install**.

## Add npm to Environment Variables - WINDOWS ONLY

1) Press **Start** and search for **Environment Variables**.
2) Click **Edit the system environment variables**.
3) In the window that opens, click **Environment Variables**.

<img src = "Assets/System_Properties.png" width="40%" >

4) Under **User variables for [your username]**, find **Path**, select it, and click **Edit...**.

<img src = "Assets/User_Variables.png" width="40%" >

5) Click **New**, and add **C:\Users\\[your username]\\.npm-global**.

<img src = "Assets/Environment_Variables.png" width="40%" >

6) Click **OK** on all open windows.

## Install Edge Impulse Data Forwarder

1) If Command Prompt/Terminal is open, close and reopen it.
2) Run `npm install -g edge-impulse-cli`
3) If there are no errors, run `edge-impulse-data-forwarder --version`.
4) If this command works, you have successfully installed the data forwarder! Now, run `edge-impulse-data-forwarder`.
5) Follow the instructions on screen to log into your Edge Impulse account.

*NOTE: To log out of your account in the data forwarder, run `edge-impulse-data-forwarder --clean`*
