# Data Cleaning

Real-world data is messy, and so the next step is to clean and organize your data before training your model.

For stall detection, we recommend using the following labels: 

- **Idle** - The motor is running unobstructed (no load). Current draw is relatively stable.
- **Pre-stall** - The motor is still running but is starting to experience more resistance. Current draw is increasing.
- **Stall** - The motor has completely stopped and current draw is stable.

Use Edge Impulse's **Crop** and **Split** features to divide each motor run into smaller segments. Try to create segments that are distinguishable and clearly represent the different states. The most important section to preserve is the pre-stall to stall transition, since the model needs to learn how to recognize this behaviour.

To rename a segment, click on the three dots to the right of it and select **Rename**. 

Data cleaning also involves removing bad data. Look out for:
- Small, quick spikes caused by noise (usually a few milliseconds wide). These can be cropped out of training data. Also, remove samples that have too much noise.

- Bad/incomplete runs. For example, if you were not able to fully stop the motor during a run and all the data appears to be pre-stall. 
- Incorrect labels. Check that each label matches the shape of the data.
