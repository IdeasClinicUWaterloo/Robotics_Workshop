Next, we need to set up the Edge Impulse Data Forwarder. This is a command-line tool that allows us to send sensor data from a device to Edge Impulse Studio, where the data can be collected, cleaned, and labelled to create training datasets for our machine learning models. Follow the steps below to install the required packages.

## Install Python 3
1) On Windows, open Command Prompt and run `python --version`. On Mac, open Terminal and run `python3 --version`. If you have Python 3 installed, you will see something like `Python 3.12.6`
2) If you do not have [Python 3](https://www.python.org/downloads/release/python-3150rc2/) installed, click on the hyperlink, scroll down, and download the installer.
3) Open the installer and follow the steps on screen.
4) Once the installation has completed, close and reopen Command Prompt/Terminal. Then, run `python --version` or `python3 --version` again to check that it has been properly installed.

## Install Node.js
1) Open Command Prompt (on Windows) or Terminal (on Mac) and run `node --version`. If you have Node.js installed, you will see something like `v24.21.0`
2) If you do not have [Node.js](https://nodejs.org/en/download/current) installed, click on the hyperlink, scroll down to the prebuilt Node.js section, select your OS and architecture, and download the installer. 

    *NOTE: to check your system architecture on Windows, open Settings > System > About. Then, look at System Type. On Mac, open the Apple menu, select About This Mac, and look at the Chip item. Silicon chips use ARM64 architecture and Intel processors use x86_64.*

## Change npm Default Directory
1) On Windows, open PowerShell and run `Set-ExecutionPolicy RemoteSigned -Scope CurrentUser in PowerShell`.
2) On Windows, in PowerShell run the following commands one after another to change npm's default directory:

    `mkdir ~/.npm-global`

    `npm config set prefix '~/.npm-global'`

    `echo 'export PATH=~/.npm-global/bin:$PATH' >> ~/.profile`

    OR

    On Mac, in Terminal run the following commands one after another to change npm's default directory:

    `mkdir ~/.npm-global`

    `npm config set prefix '~/.npm-global'`

    `echo 'export PATH=~/.npm-global/bin:$PATH' >> ~/.zprofile`

## Install Edge Impulse Data Forwarder

1) If Command Prompt/Terminal is open, close and reopen it.
2) Run `npm install -g edge-impulse-cli`
3) If there are no errors, run `edge-impulse-data-forwarder --version`.
4) If this command works, you have succesfully installed the data forwarder! Now, run `edge-impulse-data-forwarder`.
4) Follow the intructions on screen to log into your Edge Impulse account.

*NOTE: To log out of your account in the data forwarder, run `edge-impulse-data-forwarder --clean`*
